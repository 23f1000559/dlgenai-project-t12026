```python
#noise files 
noise_files = glob.glob("/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup/ESC-50-master/audio/*.wav")
```

```python
# custom dataset
class MyDataset(Dataset):
    def __init__(self,root_dir,noise_files,sr=16000,duration = 10):
        self.root_dir = root_dir
        self.noise_files = noise_files
        self.sr = sr
        self.max_len =sr*duration
        self.labels =[]
        self.dirs = []
        self.stems=['bass.wav', 'drums.wav', 'other.wav', 'vocals.wav']
        # collecting song directories and their genre
        for label , folder in enumerate(os.listdir(root_dir)):
            folder_path = os.path.join(root_dir,folder)
            for song in os.listdir(folder_path):
                song_path = os.path.join(folder_path,song)
                self.dirs.append(song_path)
                self.labels.append(label)
        # group song with their genres
        self.genre_dirs = {}
        for d,l in zip(self.dirs,self.labels):
            self.genre_dirs.setdefault(l,[]).append(d)
    # length adjusting
    def pad_truncate(self,audio):
        if len(audio) > self.max_len :
            audio = audio[:self.max_len]
        else: 
            audio = np.pad(audio,(0,(self.max_len-len(audio))))
        return audio
    # add noise function
    def add_noise(self,audio):
        noise,_ = librosa.load(random.choice(self.noise_files),sr=self.sr)
        noise = self.pad_truncate(noise)
        noise_level = np.random.uniform(0.05,0.3)
        return audio + noise*noise_level
    def __len__(self):
        return len(self.dirs)
    def __getitem__(self,idx):
        # different song dir of a genre for each stem
        same_genre_dirs= self.genre_dirs[self.labels[idx]]
        stem_audio = []
        for stem in self.stems:
            ran_dir = random.choice(same_genre_dirs)
            stem_path = os.path.join(ran_dir,stem)
            audio,_ = librosa.load(stem_path,sr=self.sr)
            audio = self.pad_truncate(audio)
            stem_audio.append(audio)
        # stems mix with random weights
        weights = np.random.uniform(0.5,1.5,4)
        mixed = sum(w*a for w,a in zip(weights,stem_audio)) 
        # Normalize
        mixed = mixed / (np.max(np.abs(mixed)) + 1e-6)
        #tempo stretch 
        if np.random.rand()<0.5:
            rate = np.random.uniform(0.85,1.15)
            mixed = librosa.effects.time_stretch(mixed,rate=rate)
            mixed = self.pad_truncate(mixed)
        # add noise
        mixed = self.add_noise(mixed)
        # melspectogram
        mel = librosa.feature.melspectrogram(
            y=mixed,n_mels=128,sr=self.sr,n_fft=1024,hop_length=512
        )
        mel_db = librosa.power_to_db(mel)
        return torch.tensor(mel_db,dtype=torch.float32).unsqueeze(0), self.labels[idx]

```

```python
root_dir = "/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup/genres_stems"
```

```python
#dataset instance
dataset = MyDataset(root_dir,noise_files)
# train val split
train_size = int(0.8* len(dataset))
val_size = len(dataset)- train_size
train_dataset,val_dataset = random_split(dataset,[train_size,val_size])
```

```python
train_loader = DataLoader(train_dataset,batch_size=32,shuffle=True)
val_loader = DataLoader(val_dataset,batch_size=32,shuffle=False)
```

## Model

```python
# Start a new wandb run 
run = wandb.init(
    # Set the wandb project where this run will be logged.
    project="23f1000559-t12026",
    entity ="23f1000559-indian-institute-of-technology-madras",
    # Track hyperparameters and run metadata.
    config={
        "epochs": 15,
        "batch_size": 32,
        "learning_rate": 0.0003,
        "model": "custom-cnn-v7"
    },
)
```

```python
## cnn architecture
class cnn(nn.Module):
    def __init__(self):
        super().__init__()
        self.c1 = nn.Conv2d(1,32,kernel_size=3,padding=1)
        self.c2 = nn.Conv2d(32,64,kernel_size=3,padding=1)
        self.c3 = nn.Conv2d(64,128,kernel_size=3,padding=1)
        self.c4 = nn.Conv2d(128,256,kernel_size=3,padding=1)
        self.bn1 = nn.BatchNorm2d(32)
        self.bn2 = nn.BatchNorm2d(64)
        self.bn3 = nn.BatchNorm2d(128)
        self.bn4 = nn.BatchNorm2d(256)
        self.pool = nn.MaxPool2d(2)
        self.drop = nn.Dropout(0.3)
        self.ap = nn.AdaptiveAvgPool2d((1,1))
        self.fc1 = nn.Linear(256,512)
        self.fc2 = nn.Linear(512,10)
        self.relu = nn.ReLU()
    def forward(self,x):
        x = self.relu(self.bn1(self.c1(x)))
        x = self.relu(self.bn2(self.c2(x)))
        x = self.pool(x)
        x = self.relu(self.bn3(self.c3(x)))
        x = self.relu(self.bn4(self.c4(x)))
        x = self.pool(x)
        x = self.ap(x)
        x = torch.flatten(x,1)
        x = self.relu(self.fc1(x))
        x = self.drop(x)
        x = self.fc2(x)
        return x
```

```python
# initalising model , loss , optimiser
device = "cuda" if torch.cuda.is_available() else "cpu"
model = cnn().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.AdamW(model.parameters(),lr=0.0003,weight_decay=1e-4)
```

## Training Loop

```python
num_epochs = 15
for epoch in range(num_epochs):
    total_loss = 0
    model.train()
    for x,y in train_loader:
        x =x.to(device)
        y = y.to(device)
        output=model(x)
        loss = criterion(output,y)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_loss +=loss.item()
    avg_loss = loss/len(train_loader)
    #Log loss to wandb
    wandb.log({"epoch": epoch+1, "train_loss": avg_loss})
    print(f"Epoch {epoch+1}/{num_epochs}, Loss: {avg_loss:.4f}")
```

## Validation Loop

```python
model.eval()
total_val_loss =0
all_preds=[]
all_labels = []
with torch.no_grad():
    for x_val,y_val in val_loader:
        x_val = x_val.to(device)
        y_val = y_val.to(device)
        output= model(x_val)
        loss = criterion(output, y_val)
        total_val_loss += loss.item()
        # store for f1
        all_preds.extend(output.argmax(dim=1).cpu().numpy())
        all_labels.extend(y_val.cpu().numpy())
avg_val_loss = total_val_loss / len(val_loader)
val_macro_f1 = f1_score(all_labels,all_preds,average="macro")
print(f"Val Loss: {avg_val_loss:.4f}, Val Macro F1: {val_macro_f1:.4f}")
#log to wandb
wandb.log({"val_loss": avg_val_loss, "val_macro_f1": val_macro_f1})
        
```

## Prediction

```python
test_df = pd.read_csv("/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup/test.csv")
```

```python
model.eval()
predicted_genres=[]
root = "/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup"
for file_path in test_df["filename"]:
    final_path = os.path.join(root,file_path)
    audio,_ = librosa.load(final_path,sr=16000)
    mel = librosa.feature.melspectrogram(
            y=audio,n_mels=128,sr=16000
        )
    mel_db = librosa.power_to_db(mel)
    tensor = torch.tensor(mel_db).unsqueeze(0).unsqueeze(0).float().to(device)
    with torch.no_grad():
        output = model(tensor)
        pred = output.argmax(dim=1).item()
        predicted_genres.append(pred)
        
```
